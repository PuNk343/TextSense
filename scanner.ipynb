```python
with open("sample_job.txt", "r", encoding="utf-8") as f:
    job_text = f.read()
with open("sample_resume.txt", "r", encoding="utf-8") as f:
    resume_text= f.read()`
```

```python
import re

def preprocess(text):
    text = text.lower()
    text= re.sub(r'[^a-z\s]', '', text)
    return text.split()
```

```python
def remove_stopwords(words):
    stopwords = {"the","and","for","with","that","this","are","was","were",
        "have","has","had","you","your","from","they","their",
        "will","can","all","our","not","but","any","may","into",
        "using","use","used"}
    return(w for w in words if w not in stopwords and len(w)>3)
```

```python
job_words= set(remove_stopwords(preprocess(job_text)))
resume_words= set(remove_stopwords(preprocess(resume_text)))
```

```python
missing = job_words - resume_words
```

```python
print("Missing Keywords:\n")
for word in sorted(missing):
    print(word)

print(f"\nTotal Missing: {len(missing)}")
```
